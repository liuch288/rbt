## FsResultDB 简单试用

```python
import datetime
import pandas as pd
from rbt.result_db.fs_result_db import FsResultDB
```

### 1. 初始化

```python
db = FsResultDB(root_path="./test_fs_store", frequency="tick")
sym = "IF2401"
date = datetime.date(2024, 1, 15)
```

### 2. 构造示例数据
列名格式必须是 `{factor_name}__{indicator}`（双下划线分隔）

```python
df = pd.DataFrame({
    "MdDMU__mid": [100.0, 100.5, 101.0, 100.8, 101.2],
    "MdDMU__bid": [99.9, 100.4, 100.9, 100.7, 101.1],
    "MdDMU__ask": [100.1, 100.6, 101.1, 100.9, 101.3],
    "TrendDMU_v0__score": [0.1, 0.3, 0.5, 0.4, 0.6],
}, index=pd.date_range("2024-01-15 09:30:00", periods=5, freq="s", name="ts"))
df
```

### 3. 保存数据

```python
db.save_data(sym=sym, date=date, new_data=df)
print("保存成功")
```

### 4. 查看已有因子

```python
db.get_existing_factors(sym=sym, date=date)
```

### 5. 读取全部数据

```python
db.get_data(sym=sym, date=date)
```

### 6. 按因子名筛选读取

```python
db.get_data(sym=sym, date=date, factors="MdDMU")
```
